# 01 · Je eigen werkomgeving inrichten

In dit practicum onderzoek je straks de gewichten en voederopname van biggen. Voordat je met echte meetgegevens werkt, richt je eerst je eigen werkplek in. Zo hoef je geen software op je laptop te installeren en raakt jouw werk niet door elkaar met dat van anderen.

In dit notebook doorloop je de volledige basisroutine van het werken in de cloud:
1. Je maakt een eigen **branch** op GitHub, zodat je een persoonlijke werkversie hebt binnen het gedeelde project.
2. Je start daarop een **Codespace** (een kant-en-klare programmeeromgeving in je browser) en controleert of Python werkt.
3. Je stelt jouw **rol** in (`gewichten` of `voederopname`) voor het vervolg van het practicum.
4. Je oefent met bestanden opslaan op twee verschillende manieren: een tijdelijke **werknotitie** die alleen op je Codespace blijft staan (net zoals je later doet met grotere databestanden) en een klein **controlebestand** met jouw instellingen dat je wel verstuurt naar GitHub.
5. Je oefent met **opslaan, committen en pushen** en leert hoe je een volgende keer verderwerkt in dezelfde Codespace.

**Nodig:** een GitHub-account met schrijfrechten op deze repository, een browser en toegang tot GitHub Codespaces.

**Route:** eigen branch → Codespace openen → omgeving controleren → rol instellen → bestanden opslaan (`werk` vs. `inlevering`) → opslaan, committen en pushen → stoppen en hervatten.

In deze versie zijn de codecellen en modelantwoorden al ingevuld. De voorbeelduitvoer hieronder is gemaakt met `student/proefdocent`. Voer alle codecellen zelf opnieuw uit zodat jouw eigen branch en instellingen verschijnen.

## 3. Het notebook en de Python-kernel openen

**Uitvoeren in: de Codespace-editor.**

Open links in **Explorer** (de bestandenlijst) het bestand `docent/notebooks/01_werkomgeving_inrichten.ipynb`. *(Op de gewone GitHub-website kun je een notebook alleen bekijken; pas in Codespaces kun je de Python-code uitvoeren.)*

Een notebook bestaat uit **tekstcellen** en **codecellen**. De **kernel** is het Python-programma op de achtergrond dat je code uitvoert en variabelen in het geheugen bewaart.

1. Klik rechtsboven op **Select Kernel** en kies de meegeleverde **Python 3.12**-omgeving (`/usr/local/bin/python`).
2. Voer de codecellen één voor één van boven naar beneden uit met **Shift + Enter** (of het afspeelknopje naast de cel).
3. Wacht tot een cel klaar is voordat je doorgaat. Gebruik **Run All** liever niet, omdat er tussen de codecellen ook opdrachten en controles staan die je zelf uitvoert.

De volgende cel controleert of je vanuit de juiste projectmap werkt en of jouw eigen `student/…`-branch actief is.

```python
from pathlib import Path
import subprocess
import re

# Git kent de hoofdmap, ook als het notebook in een onderliggende map staat.
resultaat = subprocess.run(
    ["git", "rev-parse", "--show-toplevel"],
    capture_output=True, text=True
)
if resultaat.returncode != 0:
    raise RuntimeError("Open dit notebook vanuit de repository in Codespaces.")
hoofdmap = Path(resultaat.stdout.strip())

resultaat = subprocess.run(
    ["git", "branch", "--show-current"],
    cwd=hoofdmap, capture_output=True, text=True, check=True
)
branch = resultaat.stdout.strip()
if not re.fullmatch(r"student/[a-z0-9][a-z0-9_-]*", branch):
    raise RuntimeError(
        "Maak of selecteer eerst je eigen branch: student/voornaam_achternaam. "
        "Voer daarna deze cel opnieuw uit."
    )

deelnemerscode = branch.removeprefix("student/")
print("Repositorymap:", hoofdmap)
print("Actieve branch:", branch)
print("Deelnemerscode:", deelnemerscode)
```

### Software en pakketten controleren

Python werkt met uitbreidingen (**pakketten**). In latere notebooks gebruik je `paho-mqtt` (voor MQTT-berichten), `requests` (voor API's), `pandas` (voor tabellen) en `matplotlib` (voor grafieken). Ondersteuning voor SQLite zit standaard al in Python.

De cel hieronder controleert of Python en al deze pakketten klaarstaan in jouw Codespace. De versienummers mogen iets afwijken van het voorbeeld.

```python
import sys
import platform
import sqlite3
from importlib.metadata import version, PackageNotFoundError

print("Python:", platform.python_version())
print("Besturingssysteem:", platform.system())
print("Pythonprogramma:", sys.executable)
print("SQLite:", sqlite3.sqlite_version)

ontbrekend = []
for pakket in ["paho-mqtt", "requests", "pandas", "matplotlib", "ipykernel"]:
    try:
        print(f"{pakket}: {version(pakket)}")
    except PackageNotFoundError:
        ontbrekend.append(pakket)
if ontbrekend:
    raise RuntimeError(
        "Ontbrekende pakketten: " + ", ".join(ontbrekend)
        + ". Controleer de kernel en de opbouw van de Codespace."
    )
print("De benodigde pakketten zijn beschikbaar.")
```

`rol` is een **variabele**: een naam waaronder Python een waarde onthoudt in het geheugen. Pas wanneer je de codecel **uitvoert**, neemt Python de nieuwe waarde over. Alleen de tekst in de cel aanpassen is dus niet genoeg.

## 6. Tijdelijke werkbestanden bewaren zonder Git (`gegevens/werk/`)

**Uitvoeren in: het notebook en de Codespace-editor.**

Niet elk bestand dat je in je Codespace gebruikt of aanmaakt, moet naar GitHub worden doorgestuurd. Daarom gebruikt dit project twee aparte mappen:

- **`gegevens/werk/`** is voor tijdelijke bestanden en grotere databestanden waarmee je rekent. Deze map staat in `.gitignore`: Git slaat alle bestanden in deze map over, zodat de repository overzichtelijk blijft en je nooit per ongeluk voorlopige databanken naar GitHub stuurt. De bestanden blijven wel gewoon op de schijf van je Codespace staan.
- **`gegevens/inlevering/`** is voor kleine, definitieve resultaten die je **wél** bewaart met Git en naar GitHub pusht.

Voordat je in latere notebooks met echte databanken werkt, oefen je dat principe hier met een simpel tekstbestand (`werknotitie.txt`). Zo controleer je straks zelf dat je het bestand in de editor kunt aanpassen en teruglezen met Python, maar dat het tegelijk *niet* wordt meegenomen naar GitHub.

De cel hieronder maakt `werknotitie.txt` aan als het nog niet bestaat. Bestaat het al, dan leest Python alleen de inhoud zonder jouw tekst te overschrijven.

```python
werkmap = hoofdmap / "gegevens" / "werk"
werkmap.mkdir(parents=True, exist_ok=True)
notitiepad = werkmap / "werknotitie.txt"

if not notitiepad.exists():
    notitiepad.write_text(
        "Deze notitie staat in mijn Codespace, niet op mijn laptop.\n",
        encoding="utf-8"
    )
    print("Nieuwe werknotitie aangemaakt.")
else:
    print("Bestaande werknotitie behouden.")

print("Bestand:", notitiepad.relative_to(hoofdmap))
print(notitiepad.read_text(encoding="utf-8"))
```

## 7. Een controlebestand maken om in te leveren (`gegevens/inlevering/`)

**Uitvoeren in: het notebook.**

In de vorige stap zag je hoe je bestanden lokaal in je Codespace bewaart zonder ze naar GitHub te sturen. Nu doe je het omgekeerde: je maakt in `gegevens/inlevering/` een klein bestand aan dat je in stap 8 wél gaat committen en pushen.

Dit bestand (`gegevens/inlevering/01_werkplekcontrole.json`) verzamelt jouw basisgegevens (naam, branch en gekozen rol) in **JSON-formaat**: een overzichtelijke tekststructuur met sleutel-waardeparen (zoals `"rol": "gewichten"`). Zo heb je straks meteen een concreet bestand om op GitHub te controleren of jouw instellingen goed zijn doorgekomen.

Voer je de cel opnieuw uit nadat je je rol hebt aangepast, dan wordt dit controlebestand automatisch bijgewerkt.

```python
import json

inlevermap = hoofdmap / "gegevens" / "inlevering"
inlevermap.mkdir(parents=True, exist_ok=True)
controlepad = inlevermap / "01_werkplekcontrole.json"

controle = {
    "deelnemerscode": deelnemerscode,
    "branch": branch,
    "rol": rol,
    "bronbestand": bronbestand,
    "werknotitie_aanwezig": notitiepad.is_file(),
}
controlepad.write_text(
    json.dumps(controle, ensure_ascii=False, indent=2) + "\n",
    encoding="utf-8"
)
teruggelezen = json.loads(controlepad.read_text(encoding="utf-8"))
assert teruggelezen == controle, "Het opgeslagen controlebestand wijkt af."
print("Opgeslagen:", controlepad.relative_to(hoofdmap))
print(json.dumps(teruggelezen, ensure_ascii=False, indent=2))
```

## 8. Opslaan, committen en pushen

**Uitvoeren in: de Codespace-editor en daarna op de GitHub-website.**

Om je werk veilig te bewaren en op GitHub zichtbaar te maken, doorloop je altijd deze drie stappen:

| Handeling | Wat gebeurt er? | Waar staat het resultaat? |
|---|---|---|
| **Opslaan** | De gewijzigde bestanden worden naar de schijf geschreven. | In de Codespace |
| **Committen** | Git legt een uitgekozen set wijzigingen vast als een vast versiepunt. | In de Git-geschiedenis van de Codespace |
| **Pushen** | De nieuwe commits worden naar jouw branch op GitHub verstuurd. | In de gedeelde GitHub-repository |

Een commit is dus nog geen upload naar GitHub. Pas na **pushen** is jouw nieuwste werk ook op de GitHub-website zichtbaar.

**Waarom altijd deze volgorde (`Save All → Commit → Push`)?**
- **Als je niet eerst opslaat (`Save All`):** Git kijkt alleen naar bestanden op de schijf, niet naar openstaande maar onopgeslagen tabbladen in de editor. Zonder opslaan commit en push je dus een oudere versie (of ziet Source Control je laatste aanpassingen niet).
- **Als je alleen opslaat, maar niet commit of pusht:** je werk staat wel op de schijf van jouw Codespace, maar heeft nog geen vast versiepunt in Git en staat nog **niet** op GitHub. De docent ziet je werk dan niet, en als de Codespace later verdwijnt ben je het kwijt.
- **Als je wel commit, maar niet pusht:** je hebt een versiepunt gemaakt in jouw Codespace, maar het is nog niet naar GitHub verstuurd.

1. Sla eerst al je openstaande bestanden op via **File → Save All**.
2. Klik links in de zijbalk op het icoon voor **Source Control**.
3. Bekijk de gewijzigde bestanden onder **Changes**. Beweeg over dit notebook en klik op het **+**-teken naast dat bestand. Doe hetzelfde bij `gegevens/inlevering/01_werkplekcontrole.json`. Dit heet **stagen**: je kiest zo bewust welke bestanden je in de commit opneemt (klik dus niet blind op het algemene **+**-teken bij *Changes*, zodat je geen ongewenste bestanden meestuurt).
4. `werknotitie.txt` uit `gegevens/werk/` staat niet in deze lijst, omdat Git die map overslaat.
5. Typ bovenaan een korte beschrijving, bijvoorbeeld: `Werkplek gecontroleerd en rol vastgelegd`.
6. Klik op **Commit** en daarna op **Push** (of **Publish Branch** als dit de eerste keer is dat je jouw branch naar GitHub stuurt).

> **Tip:** Blijf werken op jouw eigen `student/…`-branch in deze repository en open geen *pull request* naar de basisbranch. Vraagt GitHub toch om een fork of een andere repository, stop dan en laat eerst je schrijfrechten controleren.

**Vraag 4.** Waarom blijft `werknotitie.txt` wél behouden als je dezelfde Codespace stopt en hervat, terwijl je een Python-variabele zoals `rol` toch weer opnieuw moet instellen door de cellen uit te voeren?

**Modelantwoord 4**

`werknotitie.txt` staat op de schijf van je Codespace en blijft bewaard wanneer je diezelfde Codespace stopt en later weer hervat. Variabelen zoals `rol` bestaan alleen in het werkgeheugen van de draaiende Python-kernel. Zodra je de Codespace stopt (of de kernel herstart), wordt dat geheugen gewist en moet je de codecellen opnieuw uitvoeren om de variabelen weer te vullen.

## 10. Problemen oplossen

| Wat zie je? | Wat kun je doen? |
|---|---|
| Je kunt geen eigen branch maken | Controleer of je bent ingelogd op het juiste GitHub-account en laat je schrijfrechten op deze repository controleren. |
| Bestanden van dit practicum ontbreken op je branch | Controleer of je bij het maken van je branch vertrokken bent vanaf de basisbranch `main`. |
| Je kunt codecellen niet uitvoeren | Controleer of je in GitHub Codespaces werkt en niet op de gewone GitHub-website. |
| De branchcontrole geeft een foutmelding | Maak of selecteer eerst jouw eigen `student/…`-branch en voer de cel opnieuw uit. |
| `NameError` bij het uitvoeren van een cel | Een variabele uit een eerdere cel is nog niet aangemaakt. Voer de codecellen van boven naar beneden in volgorde uit. |
| Een pakket ontbreekt | Kies rechtsboven de meegeleverde Python-kernel (`Python 3.12`) of wacht tot het opbouwen van de Codespace helemaal klaar is. |
| `werknotitie.txt` verschijnt niet bij Source Control | Dat klopt: de map `gegevens/werk/` wordt expres door Git overgeslagen. |
| Git vraagt om naam of e-mailadres bij het committen | Stel eenmalig je Git-naam en e-mailadres in (gebruik eventueel je GitHub-noreplyadres) of vraag hulp. |
| Push wordt geweigerd | Controleer of je op jouw eigen branch werkt. Gebruik geen *force push*, maar vraag hulp met de foutmelding erbij. |
| Codespaces meldt een quota- of limietprobleem | Stop ongebruikte Codespaces via [github.com/codespaces](https://github.com/codespaces) of vraag hulp bij het controleren van je Codespaces-tegoed. |

Vraag je om hulp, vermeld dan altijd bij welke sectie het gebeurt en welke foutmelding je krijgt. Deel nooit wachtwoorden of toegangstokens.
