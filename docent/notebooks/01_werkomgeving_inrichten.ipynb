## 3. Het notebook en de Python-kernel openen

**Uitvoeren in: de Codespace-editor.**

Links staat **Explorer**, de bestandenlijst. Open `docent/notebooks/01_werkomgeving_inrichten.ipynb`. Een weergave op de gewone GitHub-website is alleen een voorbeeldweergave: daar kun je de Python-cellen niet uitvoeren.

Een notebook bestaat uit tekstcellen en codecellen. De **kernel** is het Python-proces dat de codecellen uitvoert en variabelen in het geheugen bewaart. Kies rechtsboven **Select Kernel**, vervolgens de Python-omgeving met Python 3.12. Als er meerdere mogelijkheden zijn, kies de meegeleverde omgeving met uitvoerbaar bestand `/usr/local/bin/python`.

Voer de codecellen van boven naar beneden uit met **Shift+Enter** of het afspeelknopje naast de cel. Wacht tot een cel klaar is voordat je verdergaat. Een foutmelding is uitvoer van die cel; lees die voordat je de volgende uitvoert. **Run All** is niet de eerste werkwijze: er staan ook handmatige stappen en controles tussen de cellen.

De volgende cel zoekt de hoofdmap van de repository. Ze werkt ook als de kernel vanuit de map met docentnotebooks is gestart. Daarna leest ze de actieve Git-branch. Er worden nog geen bestanden geschreven.

```python
from pathlib import Path
import subprocess
import re

# Git kent de hoofdmap, ook als het notebook in een onderliggende map staat.
resultaat = subprocess.run(
    ["git", "rev-parse", "--show-toplevel"],
    capture_output=True, text=True
)
if resultaat.returncode != 0:
    raise RuntimeError("Open dit notebook vanuit de repository in Codespaces.")
hoofdmap = Path(resultaat.stdout.strip())

resultaat = subprocess.run(
    ["git", "branch", "--show-current"],
    cwd=hoofdmap, capture_output=True, text=True, check=True
)
branch = resultaat.stdout.strip()
if not re.fullmatch(r"student/[a-z0-9][a-z0-9-]*", branch):
    raise RuntimeError(
        "Maak of selecteer eerst je eigen branch: student/jouw-deelnemerscode. "
        "Voer daarna deze cel opnieuw uit."
    )

deelnemerscode = branch.removeprefix("student/")
print("Repositorymap:", hoofdmap)
print("Actieve branch:", branch)
print("Deelnemerscode:", deelnemerscode)
```

<div style="background-color:#e8f5e9;color:#1b5e20;border-left:5px solid #2e7d32;padding:14px;margin:12px 0;border-radius:4px;"><strong>Interpretatie</strong><br>Je ziet een map op de cloudcomputer, doorgaans onder <code>/workspaces/</code>, en jouw eigen branch. De deelnemerscode wordt uit die branchnaam afgeleid. Een fout over de branch los je op door eerst stap 1 of de route voor een bestaande Codespace uit te voeren; wijzig de controlecode niet.</div>

### Controle van de software

Python kan worden uitgebreid met **pakketten**. In de vervolgoefeningen verzorgt `paho-mqtt` de MQTT-verbindingen, haalt `requests` API-antwoorden op, verwerkt `pandas` tabellen en maakt `matplotlib` grafieken. SQLite-ondersteuning zit al in Python.

De cel hieronder controleert of deze onderdelen beschikbaar zijn. De controle opent geen broker en vraagt nog geen API op. De versienummers kunnen verschillen van een voorbeeld; het gaat hier om een werkende installatie.

```python
import sys
import platform
import sqlite3
from importlib.metadata import version, PackageNotFoundError

print("Python:", platform.python_version())
print("Besturingssysteem:", platform.system())
print("Pythonprogramma:", sys.executable)
print("SQLite:", sqlite3.sqlite_version)

ontbrekend = []
for pakket in ["paho-mqtt", "requests", "pandas", "matplotlib", "ipykernel"]:
    try:
        print(f"{pakket}: {version(pakket)}")
    except PackageNotFoundError:
        ontbrekend.append(pakket)
if ontbrekend:
    raise RuntimeError(
        "Ontbrekende pakketten: " + ", ".join(ontbrekend)
        + ". Controleer de kernel en de opbouw van de Codespace."
    )
print("De benodigde pakketten zijn beschikbaar.")
```

## 6. Een werkbestand maken en terugvinden

De map `gegevens/werk/` is bestemd voor tijdelijke werkgegevens. De map `gegevens/inlevering/` is bestemd voor kleine, bewust gekozen resultaten die je met Git bewaart. In de repository staat welke bestanden Git overslaat in `.gitignore`. Een overgeslagen bestand blijft gewoon op de schijf staan; het wordt alleen niet automatisch meegenomen in versiebeheer.

De volgende cel maakt een werknotitie aan **als die nog niet bestaat**. Bij opnieuw uitvoeren wordt een bestaande notitie gelezen, niet vervangen. Zo kun je testen dat een zelf aangebrachte wijziging blijft bestaan.

```python
werkmap = hoofdmap / "gegevens" / "werk"
werkmap.mkdir(parents=True, exist_ok=True)
notitiepad = werkmap / "werknotitie.txt"

if not notitiepad.exists():
    notitiepad.write_text(
        "Deze notitie staat in mijn Codespace, niet op mijn laptop.\n",
        encoding="utf-8"
    )
    print("Nieuwe werknotitie aangemaakt.")
else:
    print("Bestaande werknotitie behouden.")

print("Bestand:", notitiepad.relative_to(hoofdmap))
print(notitiepad.read_text(encoding="utf-8"))
```

## 7. Een klein inleverbestand bewaren

Een **JSON-bestand** bevat gegevens als sleutel-waardeparen. We gebruiken hier een klein controlebestand met je deelnemerscode, branch en rol. Dat is nog geen MQTT-bericht, maar wel hetzelfde soort leesbare gegevensstructuur dat later als berichtinhoud kan worden gebruikt.

De cel maakt `gegevens/inlevering/01_werkplekcontrole.json`. Bij opnieuw uitvoeren wordt dit controlebestand bewust bijgewerkt met de actuele instellingen. Er worden geen wachtwoorden, serveradressen of andere toegangsgegevens opgenomen.

```python
import json

inlevermap = hoofdmap / "gegevens" / "inlevering"
inlevermap.mkdir(parents=True, exist_ok=True)
controlepad = inlevermap / "01_werkplekcontrole.json"

controle = {
    "deelnemerscode": deelnemerscode,
    "branch": branch,
    "rol": rol,
    "bronbestand": bronbestand,
    "werknotitie_aanwezig": notitiepad.is_file(),
}
controlepad.write_text(
    json.dumps(controle, ensure_ascii=False, indent=2) + "\n",
    encoding="utf-8"
)
teruggelezen = json.loads(controlepad.read_text(encoding="utf-8"))
assert teruggelezen == controle, "Het opgeslagen controlebestand wijkt af."
print("Opgeslagen:", controlepad.relative_to(hoofdmap))
print(json.dumps(teruggelezen, ensure_ascii=False, indent=2))
```

## 8. Opslaan, committen en pushen

De drie handelingen hebben een verschillend doel:

| Handeling | Wat gebeurt er? | Waar staat het resultaat? |

Een commit is geen upload naar GitHub. Ook een branch die je op GitHub kunt zien, bevat pas je nieuwste werk nadat je hebt gepusht. De docent kan je opgeslagen GitHub-versie bekijken; niet-gepushte wijzigingen in je Codespace zijn daar niet zichtbaar.

## 9. Stoppen en op een andere dag verderwerken

**Een browser sluiten, een kernel herstarten en een Codespace stoppen zijn verschillende handelingen.** De kernel bewaart Python-variabelen in het geheugen. Bestanden staan op schijf. GitHub bewaart alleen wat ernaartoe is gepusht.

| Handeling | Wat blijft behouden? | Wat moet je daarna doen? |
|---|---|---|
| Kernel herstarten | Opgeslagen bestanden; de variabelen verdwijnen | Codecellen opnieuw in volgorde uitvoeren |
| Codespace stoppen en hervatten | Opgeslagen bestanden in dezelfde Codespace | Notebook openen en zo nodig de kernel opnieuw starten |
| Nieuwe Codespace maken | De gepushte repositoryversie wordt opnieuw opgehaald | Niet-gepushte werkbestanden zijn daar niet automatisch beschikbaar |
| Codespace verwijderen | De Codespace zelf en niet-geëxporteerde bestanden verdwijnen | Verderwerken vanaf gepushte of apart bewaarde bestanden |

Voer aan het einde van een sessie **Save All → Commit → Push → controle op GitHub** uit. Stop daarna via het Codespaces-menu **Stop Current Codespace**, of via [github.com/codespaces](https://github.com/codespaces) en het menu naast de betreffende werkplek. Alleen de browsertab sluiten stopt de cloudcomputer niet noodzakelijk.

Hervat bij een volgende sessie via dezelfde Codespaces-lijst. Open opnieuw dit notebook en voer de cellen van boven naar beneden uit. De branchcontrole hoort nog jouw branch te tonen. Je zelf aangevulde werknotitie hoort behouden te blijven. Het inleverbestand wordt bijgewerkt met dezelfde instellingen.

Een **Rebuild Container** is een heropbouw van de softwareomgeving, niet de gewone hervatknop. Bestanden onder `/workspaces/` blijven daarbij behouden, maar zelf aangebrachte wijzigingen elders in de container kunnen verdwijnen. Codespaces kunnen na langdurige inactiviteit worden verwijderd; gebruik GitHub en gerichte exports om belangrijk werk te bewaren.

| `NameError` bij een latere cel | De kernel kent een variabele nog niet. Voer de eerdere codecellen in volgorde uit. |
| Een pakket ontbreekt | Selecteer de meegeleverde Python-kernel. Controleer of de Codespace-opbouw voltooid is; laat een bouwfout door de docent bekijken. |
| De notitie verschijnt niet in Source Control | Dat is de bedoeling: `gegevens/werk/` wordt door Git overgeslagen. |
| Git vraagt om naam of e-mailadres bij committen | Laat de docent helpen met je Git-identiteit. Gebruik eventueel je GitHub-noreplyadres. |
| Push wordt geweigerd | Controleer branch en rechten. Gebruik geen force push; laat de docent de melding bekijken. |
| Codespaces meldt een quota- of betaalprobleem | Laat de rekeninghouder het GitHub-gebruik controleren; dit staat los van DigitalOcean. |

Wanneer je hulp vraagt, vermeld de sectie en de foutmelding. Deel geen wachtwoorden of toegangstokens.
